````python
from ollama import chat
from IPython.display import display, JSON
import json
import pandas as pd
from concurrent.futures import ThreadPoolExecutor
from functools import partial
from tqdm.notebook import tqdm
import time

def get_debate_history(df, row, max_tokens=8000):
    """Get all previous speeches from the same debate"""
    try:
        # Get all speeches from same debate
        debate_rows = df[
            (df['date'] == row['date']) & 
            (df['title'] == row['title']) &
            (df['document_number'] == row['document_number'])
        ]
    
            # Sort by file_id to maintain chronological order
        debate_rows = debate_rows.sort_values('file_id')
        current_position = debate_rows.index.get_loc(row.name)
        
        # Build debate history
        history = []
        total_length = 0
        
        # Get previous speeches
        for i in range(current_position):
            prev = debate_rows.iloc[i]
            speech = (
                f"[{prev['speaker_name']} ({prev['speaker_party']})]: "
                f"{prev['speech_text']}"
            )
            
            if total_length + len(speech) > max_tokens:
                history.append("... [Earlier debate content truncated] ...")
                break
                
            history.append(speech)
            total_length += len(speech)
        
        return {
            'previous_text': "\n\n---\n\n".join(history),
            'position_info': f"Speech {current_position + 1} van {len(debate_rows)}",
            'total_speeches': len(debate_rows)
        }
    except Exception as e:
        print(f"Error in get_debate_history: {str(e)}")
        return {
            'previous_text': "",
            'position_info': "Position unknown",
            'total_speeches': 0
        }
    
  
def validate_json_format(json_str):
    """Validate if the JSON matches the required format"""
    try:
        data = json.loads(json_str)
        required_keys = {'gevonden_misvatting', 'samenvatting'}
        summary_keys = {'aantal_gevallen', 'gemiddelde_confidence', 'hoogste_confidence', 'laagste_confidence'}
        
        if not all(key in data for key in required_keys):
            return False
            
        if not all(key in data['samenvatting'] for key in summary_keys):
            return False
            
        if data['gevonden_misvatting']:
            required_misvatting_keys = {'citaat', 'toelichting', 'confidence', 'context'}
            if not all(all(key in item for key in required_misvatting_keys) 
                      for item in data['gevonden_misvatting']):
                return False
        
        return True
    except:
        return False

def process_speech(row, df, prompt_template, model="deepseek-r1:8b"):
    """Process a single speech with full debate context"""
    try:
        # Get debate history
        debate_context = get_debate_history(df, row)
        
        # Prepare context with only available fields
        context = {
            'title': row['title'],
            'document_number': row['document_number'],
            'speaker_name': row['speaker_name'],
            'speaker_party': row['speaker_party'],
            'date': row['date'],
            'category': row['category'],
            'speech_text': row['speech_text'],
            'previous_text': debate_context['previous_text'],
            'debate_position': debate_context['position_info']  # Using position_info from debate_context
        }
        
        # Format and send prompt
        formatted_prompt = prompt_template.format(**context)
        response = chat(model, messages=[{"role": "user", "content": formatted_prompt}])
        response_text = response['message']['content']
        
        # Process response
        json_start = response_text.find('{')
        json_end = response_text.rfind('}') + 1
        
        if json_start >= 0 and json_end > json_start:
            json_str = response_text[json_start:json_end]
            if validate_json_format(json_str):
                return {
                    'debate': row['title'],
                    'speaker': row['speaker_name'],
                    'position': debate_context['position_info'],
                    'result': json.loads(json_str),
                    'status': 'success'
                }
        
        return {
            'debate': row['title'],
            'speaker': row['speaker_name'],
            'position': debate_context['position_info'],
            'error': 'Invalid JSON response',
            'status': 'error'
        }
        
    except Exception as e:
        print(f"Detailed error in process_speech: {str(e)}")
        return {
            'debate': row['title'],
            'speaker': row['speaker_name'],
            'error': f"Error: {str(e)}",
            'status': 'error'
        }
        
# Read the CSV and filter for 2023-2024
df = pd.read_csv('seperated_speeches_2014-2024.tar/seperated_speeches_2014-2024/seperated_speeches_2014-2024/speeches_2023-2024.csv')

# Create sample
sample_df = df.sample(n=5, random_state=42)

# Process speeches in parallel with progress bar
results = []
with ThreadPoolExecutor(max_workers=3) as executor:
    process_func = partial(process_speech, df=df, prompt_template=prompt)
    futures = [executor.submit(process_func, row) for _, row in sample_df.iterrows()]
    
    for future in tqdm(futures, desc="Processing speeches"):
        result = future.result()
        results.append(result)
        
        # Display results as they complete
        print(f"\nAnalyzing: {result['debate']}")
        print(f"Speaker: {result['speaker']}")
        if result['status'] == 'success':
            display(JSON(result['result']))
        else:
            print(f"Error: {result.get('error', 'Unknown error')}")
        print("=" * 80)
        
        # Small delay to prevent rate limiting
        time.sleep(1)

# Display summary
print("\nProcessing Summary:")
print(f"Successfully processed: {sum(1 for r in results if r['status'] == 'success')}/5")
print(f"Failed: {sum(1 for r in results if r['status'] == 'error')}/5")
````

````python
import pandas as pd
import json
import yaml
from typing import Dict, Optional
import logging
from tqdm.notebook import tqdm
from ollama import chat
import time
from IPython.display import display, JSON

# Set up logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

class GeneticFallacyAnalyzer:
    def __init__(self, model_name: str = "deepseek-r1:8b", yaml_path: str = "geneticfallacy.yaml"):
        """Initialize analyzer with model and prompt."""
        self.model = model_name
        # Load prompt template
        with open(yaml_path, 'r', encoding='utf-8') as f:
            prompt_data = yaml.safe_load(f)
            self.prompt_template = prompt_data['prompt_template']
        
    def analyze_speech(self, text: str) -> Dict:
        """Analyze a single speech for genetic fallacies."""
        try:
            # Format prompt with speech text
            formatted_prompt = self.prompt_template.format(text=text)
            
            # Get model response
            response = chat(self.model, messages=[{
                "role": "user",
                "content": formatted_prompt
            }])
            
            # Extract and validate JSON
            result = self._extract_json(response['message']['content'])
            return result if result else self._get_empty_result()
                
        except Exception as e:
            logger.error(f"Error analyzing speech: {str(e)}")
            return self._get_empty_result()

    def _extract_json(self, text: str) -> Optional[Dict]:
        """Extract and validate JSON from model response."""
        try:
            json_start = text.find('{')
            json_end = text.rfind('}') + 1
            
            if json_start >= 0 and json_end > json_start:
                json_str = text[json_start:json_end]
                result = json.loads(json_str)
                if self._validate_result_format(result):
                    return result
            return None
        except Exception as e:
            logger.error(f"JSON extraction failed: {str(e)}")
            return None

    def _validate_result_format(self, result: Dict) -> bool:
        """Validate the analysis result format."""
        try:
            required_keys = {'gevonden_misvatting', 'samenvatting'}
            summary_keys = {
                'aantal_gevallen', 
                'gemiddelde_confidence',
                'hoogste_confidence', 
                'laagste_confidence'
            }
            return (all(key in result for key in required_keys) and 
                    all(key in result['samenvatting'] for key in summary_keys))
        except Exception:
            return False

    def _get_empty_result(self) -> Dict:
        """Return empty result structure."""
        return {
            "gevonden_misvatting": [],
            "samenvatting": {
                "aantal_gevallen": 0,
                "gemiddelde_confidence": 0.0,
                "hoogste_confidence": 0.0,
                "laagste_confidence": 0.0
            }
        }
````
